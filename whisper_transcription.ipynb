```python
import sounddevice as sd
import numpy as np
import scipy.io.wavfile as wavfile
from openai import OpenAI
import io


duration = 15
sample_rate = 16000
chunk_duration = 5

print(f"🎤 Recording for {duration} seconds... Speak continuously!")
audio = sd.rec(int(duration * sample_rate), samplerate=sample_rate, channels=1, dtype='float32')
sd.wait()
audio = audio.flatten()
print("✅ Recording complete!")

display(Audio(audio, rate=sample_rate))


chunk_size = chunk_duration * sample_rate
chunks = [audio[i:i+chunk_size] for i in range(0, len(audio), chunk_size)]

print(f"\n🔪 Split into {len(chunks)} chunks")

all_transcripts = []

for i, chunk in enumerate(chunks):
    print(f"\n🤖 Transcribing chunk {i+1}/{len(chunks)}...")
    
 
    chunk_int16 = np.int16(chunk * 32767)
    buffer = io.BytesIO()
    wavfile.write(buffer, sample_rate, chunk_int16)
    buffer.seek(0)
    buffer.name = f"chunk_{i}.wav"
    

    transcript = client.audio.transcriptions.create(
        model="whisper-1",
        file=buffer
    )
    
    all_transcripts.append(transcript.text)
    print(f"Chunk {i+1}: {transcript.text}")

print("\n📝 Complete Transcription:")
print("-" * 40)
full_text = " ".join(all_transcripts)
print(full_text)
```

```python
import sounddevice as sd
import numpy as np
import scipy.io.wavfile as wavfile
from openai import OpenAI
import io
from IPython.display import Audio, display


duration = 10
sample_rate = 16000

print(f"🎤 Recording for {duration} seconds... Speak with pauses!")
audio = sd.rec(int(duration * sample_rate), samplerate=sample_rate, channels=1, dtype='float32')
sd.wait()
audio = audio.flatten()
print("✅ Recording complete!")

display(Audio(audio, rate=sample_rate))

audio_int16 = np.int16(audio * 32767)
buffer = io.BytesIO()
wavfile.write(buffer, sample_rate, audio_int16)
buffer.seek(0)
buffer.name = "recording.wav"

print("\n🤖 Transcribing with timestamps...")
transcript = client.audio.transcriptions.create(
    model="whisper-1",
    file=buffer,
    response_format="verbose_json",
    timestamp_granularities=["segment", "word"]
)

print("\n📝 Transcription with timestamps:")
print("-" * 40)

if hasattr(transcript, 'segments'):
    for segment in transcript.segments:
        start = segment.start
        end = segment.end
        text = segment.text
        print(f"\n[{start:.2f}s - {end:.2f}s]")
        print(f"  {text}")

if hasattr(transcript, 'words'):
    print("\n🔤 Word-level timing:")
    print("-" * 40)
    for word in transcript.words:
        word_text = word.word
        start = word.start
        end = word.end
        print(f"{word_text:15} [{start:.2f}s - {end:.2f}s]")

print("\n📄 Full text:")
print(transcript.text)

print(f"\n🌍 Language detected: {transcript.language}")
print(f"⏱️ Total duration: {transcript.duration:.2f}s")
```

```python
"""Export functions for timestamped Whisper transcriptions.

Expected input: a list of segments, each with 'start', 'end' (seconds, already
offset to the full recording) and 'text'. Segments can be dicts or objects
with those attributes (like the ones returned by the OpenAI SDK).
"""

import json
from pathlib import Path


def _normalize(segments):
    """Turn dicts/SDK objects into plain dicts with start, end, text."""
    clean = []
    for seg in segments:
        if isinstance(seg, dict):
            start, end, text = seg["start"], seg["end"], seg["text"]
        else:
            start, end, text = seg.start, seg.end, seg.text
        clean.append({"start": float(start), "end": float(end), "text": text.strip()})
    return clean


def format_timestamp(seconds, srt=False):
    """Seconds -> 'HH:MM:SS' (txt) or 'HH:MM:SS,mmm' (srt)."""
    total_ms = int(round(seconds * 1000))
    hours, rem = divmod(total_ms, 3_600_000)
    minutes, rem = divmod(rem, 60_000)
    secs, ms = divmod(rem, 1000)
    if srt:
        return f"{hours:02d}:{minutes:02d}:{secs:02d},{ms:03d}"
    return f"{hours:02d}:{minutes:02d}:{secs:02d}"


def export_txt(segments, output_path):
    """Human-readable: [00:01:23 --> 00:01:27] text"""
    segments = _normalize(segments)
    with open(output_path, "w", encoding="utf-8") as f:
        for seg in segments:
            f.write(
                f"[{format_timestamp(seg['start'])} --> "
                f"{format_timestamp(seg['end'])}] {seg['text']}\n"
            )
    return output_path


def export_srt(segments, output_path):
    """SubRip subtitle format (numbered blocks, comma before milliseconds)."""
    segments = _normalize(segments)
    with open(output_path, "w", encoding="utf-8") as f:
        for i, seg in enumerate(segments, start=1):
            f.write(f"{i}\n")
            f.write(
                f"{format_timestamp(seg['start'], srt=True)} --> "
                f"{format_timestamp(seg['end'], srt=True)}\n"
            )
            f.write(f"{seg['text']}\n\n")
    return output_path


def export_json(segments, output_path, metadata=None):
    """Full structured export: metadata, full text, and per-segment timing."""
    segments = _normalize(segments)
    data = {
        "metadata": metadata or {},
        "duration": segments[-1]["end"] if segments else 0,
        "full_text": " ".join(s["text"] for s in segments),
        "segments": [
            {"id": i, **seg, "start_formatted": format_timestamp(seg["start"]),
             "end_formatted": format_timestamp(seg["end"])}
            for i, seg in enumerate(segments)
        ],
    }
    with open(output_path, "w", encoding="utf-8") as f:
        json.dump(data, f, indent=2, ensure_ascii=False)
    return output_path


def export_all(segments, base_name="transcript", output_dir="output", metadata=None):
    """Write .txt, .srt and .json files in one go."""
    out = Path(output_dir)
    out.mkdir(parents=True, exist_ok=True)
    paths = {
        "txt": export_txt(segments, out / f"{base_name}.txt"),
        "srt": export_srt(segments, out / f"{base_name}.srt"),
        "json": export_json(segments, out / f"{base_name}.json", metadata),
    }
    for kind, path in paths.items():
        print(f"Saved {kind.upper():4} -> {path}")
    return paths


if __name__ == "__main__":
    demo = [
        {"start": 0.0, "end": 3.5, "text": " Welcome everyone to the meeting."},
        {"start": 3.5, "end": 8.25, "text": " Let's start with the quarterly roadmap."},
        {"start": 3725.1, "end": 3729.9, "text": " Final item: deployment schedule."},
    ]
    export_all(demo, "demo", "output_demo", {"model": "whisper-1", "prompted": False})
```
